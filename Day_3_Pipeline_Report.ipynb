```python
pipe_age = float(input("Enter pipe age in years: "))
corrosion_rate = float(input("Enter corrosion rate (mm/year): "))
```

```python
original_thickness = 8
metal_loss = pipe_age * corrosion_rate
remaining_thickness = original_thickness - metal_loss
print("Metal loss =", metal_loss, "mm")
print("Remaining thickness =", remaining_thickness, "mm")
```

```python
if remaining_thickness >= 5:
    print("SAFE")
else:
    print("REPLACE PIPE")
```

```python
pipe_name = input("Enter pipe name: ")

print("Pipe:", pipe_name)
print("Metal loss =", metal_loss, "mm")
print("Remaining thickness =", remaining_thickness, "mm")

if remaining_thickness >= 5:
    print("Status: SAFE")
else:
    print("Status: REPLACE PIPE")
```

```python
inspection_year = int(input("Enter inspection year: "))

print("Inspection Year:", inspection_year)
print("Pipe:", pipe_name)
print("Metal loss:", metal_loss, "mm")
print("Remaining thickness:", remaining_thickness, "mm")

if remaining_thickness >= 5:
    print("Status: SAFE")
else:
    print("Status: REPLACE PIPE")
```

```python
pipe_age = 10
corrosion_rate = 0.2
metal_loss = pipe_age * corrosion_rate
original_thickness = 8
remaining_thickness = original_thickness - metal_loss
print("metal_loss =", metal_loss, "mm")
print("remaining_thickness =", remaining_thickness, "mm")
```

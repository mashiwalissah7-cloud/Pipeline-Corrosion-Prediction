# Pipeline Corrosion Prediction and Integrity Assessment System

## Project Goal

This project estimates pipeline metal loss, remaining wall thickness, corrosion risk level, and maintenance recommendation using pipe age, corrosion rate, and original wall thickness.

## Engineering Problem

Industrial pipelines can lose wall thickness over time due to corrosion. If the remaining wall thickness becomes too low, the pipeline may become unsafe and may require monitoring, repair, or replacement.

## What This Project Will Do

1. Analyze one pipeline.
2. Analyze multiple pipelines.
3. Classify risk level.
4. Create graphs.
5. Make maintenance recommendations.
6. Prepare results for a technical report.

```python
# Section 1: Single Pipeline Analysis

pipe_name = input("Enter pipeline name: ")
pipe_age = float(input("Enter pipe age (years): "))
corrosion_rate = float(input("Enter corrosion rate (mm/year): "))
original_thickness = float(input("Enter original wall thickness (mm): "))
```

```python
# Section 2: Corrosion Calculation

metal_loss = pipe_age * corrosion_rate
remaining_thickness = original_thickness - metal_loss
```

```python
# Section 3: Display Reaults

print("Pipeline:", pipe_name)
print("Pipe age:", pipe_age, "years")
print("Corrosion rate:", corrosion_rate, "mm/year")
print("Metal loss:", metal_loss, "mm")
print("Remaining thickkness:", remaining_thickness, "mm")
```

```python
# Section 4: Safety Classification

if remaining_thickness >= 5:
    print("Pipeline Status: SAFE")
else:
    print("Pipeline Status: REPLACE PIPE")
```

```python
# Section 5: Risk Level

if remaining_thickness >= 7:
    print("Risk Level: LOW")
elif remaining_thickness >= 5:
    print("Risk Level: MODERATE")
elif remaining_thickness >= 3:
    print("Risk Level: HIGH")
else:
    print("Risk Level: CRITICAL")
```

```python
# Section 6: Maintenance Recomendation

if remaining_thickness >= 7:
    print("Recommendation: Continue normal operation")
elif remaining_thickness >= 5:
    print("Recommendation: Increase inspection frequency")
elif remaining_thickness >= 3:
    print("Recommendation: Schedule repair")
else:
    print("Recommendation: Replace pipeline immediately")
```

```python
# Section 7: Final Inspection Summary

print("---------- PIPELINE INSPECTION REPORT ----------")

print("Pipeline:", pipe_name)
print("Pipe Age:", pipe_age, "years")
print("Corrosion Rate:", corrosion_rate, "mm/year")
print("Original Thickness:", original_thickness, "mm")
print("Metal Loss:", metal_loss, "mm")
print("Remaining Thickness:", remaining_thickness, "mm")

if remaining_thickness >= 7:
    risk_level = "LOW"
    recommendation = "Continue normal operation"

elif remaining_thickness >= 5:
    risk_level = "MODERATE"
    recommendation = "Increase inspection frequency"

elif remaining_thickness >= 3:
    risk_level = "HIGH"
    recommendation = "Schedule repair"

else:
    risk_level = "CRITICAL"
    recommendation = "Replace pipeline immediately"

print("Risk Level:", risk_level)
print("Recommendation:", recommendation)

print("-----------------------------------------------------")
```

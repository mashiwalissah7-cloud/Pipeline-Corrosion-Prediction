# Pipeline Corrosion Prediction and Integrity Assessment System

## Project Goal

This project estimates pipeline metal loss, remaining wall thickness, corrosion risk level, and maintenance recommendation using pipe age, corrosion rate, and original wall thickness.

## Engineering Problem

Industrial pipelines can lose wall thickness over time due to corrosion. If the remaining wall thickness becomes too low, the pipeline may become unsafe and may require monitoring, repair, or replacement.

## What This Project Will Do

1. Analyze one pipeline.
2. Analyze multiple pipelines.
3. Classify risk level.
4. Create graphs.
5. Make maintenance recommendations.
6. Prepare results for a technical report.

```python
# Section 1: Single Pipeline Analysis

pipe_name = input("Enter pipeline name: ")
pipe_age = float(input("Enter pipe age (years): "))
corrosion_rate = float(input("Enter corrosion rate (mm/year): "))
original_thickness = float(input("Enter original wall thickness (mm): "))
```

```python
# Section 2: Corrosion Calculation

metal_loss = pipe_age * corrosion_rate
remaining_thickness = original_thickness - metal_loss
```

```python
# Section 3: Display Reaults

print("Pipeline:", pipe_name)
print("Pipe age:", pipe_age, "years")
print("Corrosion rate:", corrosion_rate, "mm/year")
print("Metal loss:", metal_loss, "mm")
print("Remaining thickkness:", remaining_thickness, "mm")
```

```python
# Section 4: Safety Classification

if remaining_thickness >= 5:
    print("Pipeline Status: SAFE")
else:
    print("Pipeline Status: REPLACE PIPE")
```

```python
# Section 5: Risk Level

if remaining_thickness >= 7:
    print("Risk Level: LOW")
elif remaining_thickness >= 5:
    print("Risk Level: MODERATE")
elif remaining_thickness >= 3:
    print("Risk Level: HIGH")
else:
    print("Risk Level: CRITICAL")
```

```python
# Section 6: Maintenance Recomendation

if remaining_thickness >= 7:
    print("Recommendation: Continue normal operation")
elif remaining_thickness >= 5:
    print("Recommendation: Increase inspection frequency")
elif remaining_thickness >= 3:
    print("Recommendation: Schedule repair")
else:
    print("Recommendation: Replace pipeline immediately")
```

```python
# Section 7: Final Inspection Summary

print("---------- PIPELINE INSPECTION REPORT ----------")

print("Pipeline:", pipe_name)
print("Pipe Age:", pipe_age, "years")
print("Corrosion Rate:", corrosion_rate, "mm/year")
print("Original Thickness:", original_thickness, "mm")
print("Metal Loss:", metal_loss, "mm")
print("Remaining Thickness:", remaining_thickness, "mm")

if remaining_thickness >= 7:
    risk_level = "LOW"
    recommendation = "Continue normal operation"

elif remaining_thickness >= 5:
    risk_level = "MODERATE"
    recommendation = "Increase inspection frequency"

elif remaining_thickness >= 3:
    risk_level = "HIGH"
    recommendation = "Schedule repair"

else:
    risk_level = "CRITICAL"
    recommendation = "Replace pipeline immediately"

print("Risk Level:", risk_level)
print("Recommendation:", recommendation)

print("-----------------------------------------------------")
```

```python
# Section 8: Remaining Useful Life Prediction

minimum_safe_thickness = 5

if corrosion_rate > 0:
    remaining_useful_life = (remaining_thickness - minimum_safe_thickness) / corrosion_rate
else:
    remaining_useful_life = "Cannot calculate because corrosion rate is zero"

print("Minimum Safe Thickness:", minimum_safe_thickness, "mm")

if isinstance(remaining_useful_life, str):
    print("Remaining Useful Life:", remaining_useful_life)
else:
    print("Estimated Remaining Useful Life:", remaining_useful_life, "years")
```

```python
# Section 9: Future Thickness Prediction

print("---------- FUTURE THICKNESS PREDICTION ----------")

for year in range(1, 6):
    predicted_thickness = remaining_thickness - (corrosion_rate * year)
    print("After", year, "year(s):", predicted_thickness, "mm")

print("------------------------------------------------")
```

```python
# Section 10: Failure Year Prediction

minimum_safe_thickness = 5

future_year = 0
future_thickness = remaining_thickness

while future_thickness > minimum_safe_thickness:
    future_year += 1
    future_thickness = remaining_thickness - (corrosion_rate * future_year)

print("Pipeline will reach the minimum safe thickness after",
      future_year, "years.")
```

```python
# Section 11: Maintenance Urgency Classification

if corrosion_rate <= 0:
    maintenance_urgency = "UNKNOWN"
    urgency_reason = "Corrosion rate is zero or invalid, so urgency cannot be estimated."

elif remaining_useful_life <= 1:
    maintenance_urgency = "IMMEDIATE"
    urgency_reason = "Pipeline may reach unsafe thickness within 1 year."

elif remaining_useful_life <= 3:
    maintenance_urgency = "HIGH"
    urgency_reason = "Pipeline may reach unsafe thickness within 3 years."

elif remaining_useful_life <= 5:
    maintenance_urgency = "MODERATE"
    urgency_reason = "Pipeline may reach unsafe thickness within 5 years."

else:
    maintenance_urgency = "LOW"
    urgency_reason = "Pipeline has more than 5 years before reaching minimum safe thickness."

print("Maintenance Urgency:", maintenance_urgency)
print("Reason:", urgency_reason)
```

```python
# Section 12: Pipeline Integrity Score

thickness_ratio = remaining_thickness / original_thickness

if isinstance(remaining_useful_life, str):
    life_score = 0
else:
    life_score = remaining_useful_life / 10

if life_score > 1:
    life_score = 1

integrity_score = (0.7 * thickness_ratio + 0.3 * life_score) * 100

print("---------- PIPELINE INTEGRITY SCORE ----------")

print("Thickness Ratio:", thickness_ratio)
print("Life Score:", life_score)
print("Integrity Score:", round(integrity_score, 2), "out of 100")

if integrity_score >= 80:
    print("Integrity Category: EXCELLENT")

elif integrity_score >= 65:
    print("Integrity Category: GOOD - MONITOR CONDITION")

elif integrity_score >= 50:
    print("Integrity Category: CONCERN - PLAN MAINTENANCE")

else:
    print("Integrity Category: POOR - URGENT ACTION REQUIRED")

print("---------------------------------------------")
```

```python
# Section 13: Engineering Action Plan

print("---------- ENGINEERING ACTION PLAN ----------")

if integrity_score < 50 or maintenance_urgency == "IMMEDIATE":
    priority_level = "PRIORITY 1"
    engineering_decision = "Immediate engineering review required"
    inspection_interval = "Inspect immediately"
    action_plan = "Prepare for urgent repair or replacement"

elif integrity_score < 65 or maintenance_urgency == "HIGH":
    priority_level = "PRIORITY 2"
    engineering_decision = "High attention required"
    inspection_interval = "Inspect within 3 months"
    action_plan = "Schedule repair planning and increase monitoring"

elif integrity_score < 80 or maintenance_urgency == "MODERATE":
    priority_level = "PRIORITY 3"
    engineering_decision = "Continue operation with increased inspection frequency"
    inspection_interval = "Inspect within 6 months"
    action_plan = "Monitor corrosion growth and update remaining life prediction"

else:
    priority_level = "PRIORITY 4"
    engineering_decision = "Continue normal operation"
    inspection_interval = "Inspect during normal inspection cycle"
    action_plan = "Maintain routine corrosion monitoring"

print("Priority Level:", priority_level)
print("Engineering Decision:", engineering_decision)
print("Suggested Inspection Interval:", inspection_interval)
print("Action Plan:", action_plan)

print("---------------------------------------------")
```

```python
# Section 14: Risk-Based Inspection Assessment

print("---------- RISK-BASED INSPECTION ASSESSMENT ----------")

consequence_level = input("Enter consequence level (LOW, MEDIUM, HIGH): ").upper()

if integrity_score >= 80:
    probability_score = 1
    probability_level = "LOW"

elif integrity_score >= 65:
    probability_score = 2
    probability_level = "MODERATE"

elif integrity_score >= 50:
    probability_score = 3
    probability_level = "HIGH"

else:
    probability_score = 4
    probability_level = "CRITICAL"


if consequence_level == "LOW":
    consequence_score = 1

elif consequence_level == "MEDIUM":
    consequence_score = 2

elif consequence_level == "HIGH":
    consequence_score = 3

else:
    consequence_score = 2
    consequence_level = "MEDIUM"
    print("Invalid consequence level entered. Defaulting to MEDIUM.")


rbi_score = probability_score * consequence_score

if rbi_score <= 2:
    rbi_category = "LOW RBI PRIORITY"

elif rbi_score <= 4:
    rbi_category = "MEDIUM RBI PRIORITY"

elif rbi_score <= 8:
    rbi_category = "HIGH RBI PRIORITY"

else:
    rbi_category = "CRITICAL RBI PRIORITY"


print("Probability Level:", probability_level)
print("Probability Score:", probability_score)
print("Consequence Level:", consequence_level)
print("Consequence Score:", consequence_score)
print("RBI Score:", rbi_score)
print("RBI Category:", rbi_category)

print("-----------------------------------------------------")
```

```python
# Section 15: Final Engineering Decision Summary

print("========== FINAL ENGINEERING DECISION SUMMARY ==========")

print("Pipeline Name:", pipe_name)
print("Pipe Age:", pipe_age, "years")
print("Original Wall Thickness:", original_thickness, "mm")
print("Corrosion Rate:", corrosion_rate, "mm/year")
print("Metal Loss:", metal_loss, "mm")
print("Remaining Thickness:", remaining_thickness, "mm")

print("--------------------------------------------------------")

print("Risk Level:", risk_level)
print("Maintenance Recommendation:", recommendation)
print("Remaining Useful Life:", remaining_useful_life, "years")
print("Maintenance Urgency:", maintenance_urgency)
print("Integrity Score:", round(integrity_score, 2), "out of 100")

print("--------------------------------------------------------")

print("Risk-Based Inspection Category:", rbi_category)
print("Engineering Priority:", priority_level)
print("Engineering Decision:", engineering_decision)
print("Suggested Inspection Interval:", inspection_interval)
print("Action Plan:", action_plan)

print("========================================================")
```
